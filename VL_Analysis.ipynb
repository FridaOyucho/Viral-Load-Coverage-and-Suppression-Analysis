```python
#Filter test name
PatientLabsVL = PatientLabs[PatientLabs['TestName'].str.contains('VIRAL', case=False, na=False)]
```

```python
#Convert to datetime
PatientLabsVL['OrderedbyDate'] = pd.to_datetime(PatientLabsVL['OrderedbyDate'])
PatientLabsVL['DateSampleTaken'] = pd.to_datetime(PatientLabsVL['DateSampleTaken'])
```

```python
# Get min and max dates
print("Earliest date:", PatientLabs['OrderedbyDate'].min())
print("Latest date:", PatientLabs['OrderedbyDate'].max())
```

```python
#filter dates to August 2026
PatientLabsVL_filtered = PatientLabsVL[
    (PatientLabsVL['OrderedbyDate'] >= '2024-01-01') &
    (PatientLabsVL['OrderedbyDate'] <= '2026-08-31')
    ]
PatientLabsVL_filtered.shape
```

```python
#Confirm all dates are between Jan 2024 through August 2026
PatientLabsVL_filtered [['OrderedbyDate']].agg(['min', 'max'])
```

```python
#Filter the data to only required variables
VLData = PatientLabsVL_filtered[
    ['PatientPKHash',
     'SiteCode',
     'TestName',
     'OrderedbyDate',
     'TestResult',
     'DateSampleTaken',
     'SampleType']
                                ]
VLData.shape
```

```python
#UniqueID. Concat PatientPKHash with SiteCode
VLData['UniqueID'] = (VLData['PatientPKHash'].astype(str) +

VLData['SiteCode'].astype(str)
)


#Make UniqueID first col
cols = ['UniqueID'] + [c for c in VLData.columns if c != 'UniqueID']
VLData = VLData[cols]
```

```python
VLData
```

```python
#identify numerical and non-numerical values
total = len(VLData)
numeric = pd.to_numeric(VLData['TestResult'], errors='coerce').notna().sum()
non_numeric = pd.to_numeric(VLData['TestResult'], errors='coerce').isna().sum()

print(f"Total: {total:,}")
print(f"Numeric: {numeric:,}")
print(f"Non-numeric: {non_numeric:,}")
```

```python
#Tabulate test results
VLData.loc[
    pd.to_numeric(VLData['TestResult'], errors='coerce').isna(),
    'TestResult'
].value_counts()
```

```python
# Standardize and map to the correct values
VLData['TestResult2'] = (
    VLData['TestResult']
    .astype(str)
    .str.strip()
)

# Mapping rules
mapping = {
    'Undetectable': 0,
    'Not detected': 0,
    'NOT DETECTED': 0,
    'Trget Not Detected': 0,
    'Target Not Detected.': 0,
    'ND': 0,
    'No': 0,
    'LDL': 49,
    '<LDL': 49,
    '< 40': 39,
    '<40': 39,
    '> 10000000': 10000000,
    'Rejected': np.nan,
    'Invalid': np.nan,
    'INVALID': np.nan,
    'Yes': np.nan,
    'Failed': np.nan,
    'Detected': np.nan,
    'Detected DBS': np.nan,
    'POOR SAMPLE QUALITY': np.nan,
    'Collect New Sample': np.nan,
    'BEYOND DETECTABLE LIMIT': 0,
    '-1': np.nan,
    '': np.nan
}
# Apply mappings
VLData['TestResult2'] = VLData['TestResult2'].replace(mapping)

# Convert everything else to numeric
VLData['TestResult2'] = pd.to_numeric(
    VLData['TestResult2'],
    errors='coerce'
)
```

Identify Each Client’s Most Recent Viral Load

```python
#Sort records by OrderedByDate in descending order within each client.
VLData = VLData.sort_values(
    by=['UniqueID', 'OrderedbyDate'],
    ascending=[True, False]
)
```

```python
#Recent VL
# Remove records with missing dates
VLData = VLData.dropna(subset=['OrderedbyDate'])

# Get most recent VL per client
MostRecentVL = VLData.loc[
    VLData.groupby('UniqueID')['OrderedbyDate'].idxmax()
].copy()

# Sort
MostRecentVL = MostRecentVL.sort_values('OrderedbyDate', ascending=False)

print(MostRecentVL.shape)
```
