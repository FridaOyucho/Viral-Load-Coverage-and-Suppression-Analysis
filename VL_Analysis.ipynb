```python
# Standardize and map to the correct values
VLData['TestResult2'] = (
    VLData['TestResult']
    .astype(str)
    .str.strip()
)

# Mapping rules
mapping = {
    'Undetectable': 0,
    'Not detected': 0,
    'NOT DETECTED': 0,
    'Trget Not Detected': 0,
    'Target Not Detected.': 0,
    'ND': 0,
    'No': 0,
    'LDL': 49,
    '<LDL': 49,
    '< 40': 39,
    '<40': 39,
    '> 10000000': 10000000,
    'Rejected': np.nan,
    'Invalid': np.nan,
    'INVALID': np.nan,
    'Yes': np.nan,
    'Failed': np.nan,
    'Detected': np.nan,
    'Detected DBS': np.nan,
    'POOR SAMPLE QUALITY': np.nan,
    'Collect New Sample': np.nan,
    'BEYOND DETECTABLE LIMIT': 0,
    '-1': np.nan,
    '': np.nan
}
# Apply mappings
VLData['TestResult2'] = VLData['TestResult2'].replace(mapping)

# Convert everything else to numeric
VLData['TestResult2'] = pd.to_numeric(
    VLData['TestResult2'],
    errors='coerce'
)
```

Identify Each Client’s Most Recent Viral Load

```python
#Sort records by OrderedByDate in descending order within each client.
VLData = VLData.sort_values(
    by=['UniqueID', 'OrderedbyDate'],
    ascending=[True, False]
)
```

```python
#Recent VL
# Remove records with missing dates
VLData = VLData.dropna(subset=['OrderedbyDate'])

# Get most recent VL per client
MostRecentVL = VLData.loc[
    VLData.groupby('UniqueID')['OrderedbyDate'].idxmax()
].copy()

# Sort
MostRecentVL = MostRecentVL.sort_values('OrderedbyDate', ascending=False)

print(MostRecentVL.shape)
```

Merge the Line List with the Latest Viral Load Dataset

```python
# Format dates as datetimes
FactART_filtered['LastVisitDate'] = pd.to_datetime(FactART_filtered['LastVisitDate'])
FactART_filtered['LastVLDate'] = pd.to_datetime(FactART_filtered['LastVLDate'])
FactART_filtered['DOB'] = pd.to_datetime(FactART_filtered['DOB'])
FactART_filtered['StartARTDate'] = pd.to_datetime(FactART_filtered['StartARTDate'])

#Calculate Age
FactART_filtered['Age'] = (FactART_filtered['LastVisitDate'] - FactART_filtered['DOB']).dt.days / 365.25
```

```python
#Calculate the time elapsed (in months) since the last Viral Load test
FactART_filtered['MonthsSinceLastVL'] = (FactART_filtered['LastVisitDate'] - FactART_filtered['LastVLDate']).dt.days / 365
```

```python
#Define conditional validity windows (in months) per MOH criteria
# Pregnant/Breastfeeding = 6 months
# Children/Adolescents (<20 years) = 6 months
# General Adults = 12 months

is_pbfw = (FactART_filtered['Pregnant'] == 1) | (FactART_filtered['Pregnant'].astype(str).str.lower().str.startswith('y')) | \
          (FactART_filtered['Breastfeeding'] == 1) | (FactART_filtered['Breastfeeding'].astype(str).str.lower().str.startswith('y'))

is_child_adolescent = FactART_filtered['Age'] < 20
```

```python
# Assign the allowed window dynamically
FactART_filtered['RequiredVLWindow'] = np.where(is_pbfw | is_child_adolescent, 6, 12)
```

```python
#Evaluate if the patient has a valid VL
# A VL is valid if a test exists and fell within the patient's designated window
FactART_filtered['CalculatedHasValidVL'] = np.where(
    FactART_filtered['LastVLDate'].notna() & (FactART_filtered['MonthsSinceLastVL'] <= FactART_filtered['RequiredVLWindow']),
    1,
    0
)
```
